O que é um Node?

Um Node é uma tarefa.

```python
def responder(state: AgentState):

    pergunta = state["pergunta"]

    resposta = f"Você perguntou: {pergunta}"

    return {
        "resposta": resposta
    }
```

Criando o Graph

```python
graph = StateGraph(AgentState)
```

Adicionando o Node

```python
graph.add_node("responder", responder)
```

Criando as conexões

```python
graph.add_edge(START, "responder")
graph.add_edge("responder", END)
```

Compilar

```python
app = graph.compile()
```

Executar

```python
resultado = app.invoke({
    "pergunta": "Quanto é 10 + 20?",
    "resposta": ""
})

print(resultado)
```

Conditional Edge

Adicionando informação ao State

```python
class AgentState(TypedDict):
    pergunta: str
    resposta: str
    tipo: str
```

Criando o Node de Classificação

```python
def classificar(state: AgentState):

    pergunta = state["pergunta"].lower()

    if "quanto" in pergunta or "+" in pergunta or "*" in pergunta:
        tipo = "matematica"
    else:
        tipo = "conversa"

    return {
        "tipo": tipo
    }
```

```python
def matematica(state: AgentState):

    return {
        "resposta": "Vou utilizar uma calculadora para resolver isso 🧮"
    }
```

```python
def conversa(state: AgentState):

    return {
        "resposta": "Essa parece ser uma pergunta de conversa 💬"
    }
```

Criar o roteador

Agora precisamos de uma função que diga para onde ir

```python
def decidir_caminho(state: AgentState):

    if state["tipo"] == "matematica":
        return "matematica"

    return "conversa"
```

```python
graph = StateGraph(AgentState)

graph.add_node("classificar", classificar)
graph.add_node("matematica", matematica)
graph.add_node("conversa", conversa)
```

              classificar
              /         \
             /           \
     matematica         conversa

Conectando START

```python
graph.add_edge(START, "classificar")
```

Criando a Decisão

```python
graph.add_conditional_edges(
    "classificar",
    decidir_caminho,
    {
        "matematica": "matematica",
        "conversa": "conversa"
    }
)
```

Conectar ao END

```python
graph.add_edge("matematica", END)
graph.add_edge("conversa", END)
```

                    ▼

# LLM decidindo qual Tool usar

```python
!pip install -q -U langgraph langchain langchain-google-genai
```

```python
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langgraph.graph import StateGraph, START, END
```

Criando o modelo

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GEMINI_API_KEY,
    temperature=0
)
```

Criando a Tool

```python
@tool
def calcular(a: float, b: float) -> float:
    """Soma dois números."""
    return a + b

@tool
def multiplicar(a: float, b: float) -> float:
    """Multiplica dois números."""
    return a * b
```

```python
@tool
def consultar_produto(produto: str) -> str:
    """Consulta informações sobre um produto."""

    produtos = {
        "notebook": "Notebook Dell - R$ 4.500",
        "mouse": "Mouse Logitech - R$ 150",
        "teclado": "Teclado Mecânico - R$ 350"
    }

    return produtos.get(
        produto.lower(),
        "Produto não encontrado."
    )
```

                 🧠 LLM
                   │
       ┌───────────┼───────────┐
       ▼           ▼           ▼
   🧮 calcular  ✖️ multiplicar  📦 produto

Entregando as Tools para a LLM

```python
tools = [
    calcular,
    multiplicar,
    consultar_produto
]

llm_with_tools = llm.bind_tools(tools)
```

Testando

```python
response = llm_with_tools.invoke(
    "Quanto é 15 multiplicado por 8?"
)

print(response)
```

Criando o State do nosso agent

```python
from typing import Annotated
from langgraph.graph.message import add_messages

class AgentState(TypedDict):
    messages: Annotated[list, add_messages]
```

    🎒 STATE

    messages
      │
      ├── usuário
      ├── IA
      ├── Tool
      └── IA

```python
def chatbot(state: AgentState):

    response = llm_with_tools.invoke(
        state["messages"]
    )

    return {
        "messages": [response]
    }
```

```python
from langgraph.prebuilt import ToolNode

tool_node = ToolNode(tools)
```

    🧠 LLM
      │
      ▼
    ToolNode
      │
      ├── calcular
      ├── multiplicar
      └── consultar_produto

Criando a decisão

```python
def should_continue(state: AgentState):

    last_message = state["messages"][-1]

    if last_message.tool_calls:
        return "tools"

    return END
```

```python
graph = StateGraph(AgentState)

graph.add_node("chatbot", chatbot)
graph.add_node("tools", tool_node)
```

```python
graph.add_edge(START, "chatbot")
```

```python
graph.add_edge(START, "chatbot")
```

```python
graph.add_edge("tools", "chatbot")
```

```python
app = graph.compile()
```

```python
result = app.invoke({
    "messages": [
        {
            "role": "user",
            "content": "Quanto é 15 multiplicado por 8?"
        }
    ]
})

for message in result["messages"]:
    print(message)
```
